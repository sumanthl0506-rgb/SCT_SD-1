```python
import random

secret_number = random.randint(1, 100)
attempts = 0
print("Task 02 - Guess the number between 1 and 100")

while True:
    guess = int(input("Enter your guess: "))
    attempts += 1
    if guess == secret_number:
        print(f"You WON in {attempts} attempts! Number was {secret_number}")
        break
    elif guess < secret_number:
        print("Too Low! Try higher.")
    else:
        print("Too High! Try lower.")
```
